
# Statistical Analysis of Asphalt Concrete Mixtures


## 1. Introduction

This project investigates the mechanical properties of asphalt concrete mixtures using statistical methods. The analysis focuses on stiffness modulus, asphalt mixture type, temperature, and other material properties.

The study will use descriptive statistics, confidence intervals, and inferential statistical tests to investigate relationships and differences in the dataset.


```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load all worksheets from the Excel file
file_path = "Full Sample Data.xlsx"

sheets = pd.read_excel(file_path, sheet_name=None)

# Display worksheet names and dimensions
for name, df in sheets.items():
    print(f"{name}: {df.shape}")
```


## 2. Data Inspection

The dataset contains four worksheets: HMA2, HMA3, SMA1, and SMA2.

This section examines the column names, data types, and structure of each worksheet before data cleaning and statistical analysis.


```python

# Inspect column names and data types

for name, df in sheets.items():
    print(f"\n===== {name} =====")
    print("Shape:", df.shape)
    print("\nColumn names:")
    print(df.columns.tolist())
    print("\nData types:")
    print(df.dtypes)
```


## 3. Data Quality Assessment

This section checks all four worksheets for missing values and duplicate records.

Missing values may affect statistical calculations, while duplicate records may bias the results.

Each worksheet is checked separately before combining the datasets.


```python

# Check missing values and duplicates

for name, df in sheets.items():

    print(f"\n===== {name} =====")

    print("Total missing values:")
    print(df.isnull().sum().sum())

    print("\nMissing values by column:")
    print(df.isnull().sum())

    print("\nDuplicate rows:")
    print(df.duplicated().sum())
```


### 3.1 Data Quality Results

All four worksheets (HMA2, HMA3, SMA1, and SMA2) were checked for missing values and duplicate records.

The results showed that:

- No missing values were identified in any worksheet.
- No exact duplicate rows were found.
- Each worksheet contains 45 observations.

Therefore, no missing-value imputation or duplicate removal is required at this stage.

However, further checks of variable consistency and measurement validity are necessary before statistical analysis.



## 4. Column Consistency Assessment

Before combining the four worksheets, their column names are compared to identify any structural differences.

This step ensures that variables are aligned correctly and prevents errors during data integration.


```python

# Compare column names across worksheets

all_columns = set()

for df in sheets.values():
    all_columns.update(df.columns)

for name, df in sheets.items():
    print(f"\n===== {name} =====")

    missing_columns = all_columns - set(df.columns)

    print("Number of columns:", len(df.columns))
    print("Columns not present:", missing_columns)
```


### 4.1 Column Consistency Results

The column consistency assessment identified a structural difference between the four worksheets.

HMA2 and HMA3 each contain 16 columns, while SMA1 and SMA2 contain 15 columns.

The variable "Stab/Flow" is present in both HMA worksheets but absent from both SMA worksheets.

The remaining column names are consistent across all four worksheets.

Therefore, the 15 common variables will be used for the combined statistical analysis. The original worksheets will remain unchanged.



## 5. Dataset Integration

The four worksheets (HMA2, HMA3, SMA1, and SMA2) are combined into one dataset for statistical analysis.

A new categorical variable, `Mix_Type`, is created to identify the original asphalt mixture group.

Only the 15 common columns are retained to ensure consistency across the four worksheets. The original datasets remain unchanged.


```python

# Identify common columns
common_columns = list(sheets[next(iter(sheets))].columns)

for df in sheets.values():
    common_columns = [col for col in common_columns if col in df.columns]

# Combine worksheets and add mixture type
combined_data = pd.concat([df[common_columns].assign(Mix_Type=name) for name, df in sheets.items()],ignore_index=True)

# Check the combined dataset
print("Combined dataset shape:", combined_data.shape)

print("\nObservations by mixture type:")
print(combined_data["Mix_Type"].value_counts())

print("\nFirst five rows:")
display(combined_data.head())
```

### 5.1 Dataset Integration Results

The four worksheets were successfully combined into a single dataset containing 180 observations and 16 variables.

The combined dataset contains 15 numerical variables and one categorical variable, Mix_Type.

Each mixture type (HMA2, HMA3, SMA1, and SMA2) contains 45 observations, representing 25% of the total dataset.

The Stab/Flow variable was excluded because it was not available in the SMA worksheets.

The original worksheets remain unchanged.


## 6. Combined Dataset Validation

After combining the four worksheets, the integrated dataset is checked for missing values, duplicate records, and data types.

## 8. Descriptive Statistics

Descriptive statistics are used to summarise the numerical characteristics of the selected asphalt mixture variables.

Measures of central tendency, dispersion, and distribution are calculated to understand the physical and mechanical properties of asphalt mixtures.

These statistics provide an initial understanding of the dataset before confidence interval estimation and hypothesis testing.


```python

# Select numerical variables
numerical_data = selected_data.select_dtypes(include=np.number)

# Calculate descriptive statistics
descriptive_stats = numerical_data.describe().T

# Calculate additional statistical measures
descriptive_stats["median"] = numerical_data.median()
descriptive_stats["variance"] = numerical_data.var()
descriptive_stats["range"] = numerical_data.max() - numerical_data.min()
descriptive_stats["IQR"] = numerical_data.quantile(0.75) - numerical_data.quantile(0.25)
descriptive_stats["skewness"] = numerical_data.skew()
descriptive_stats["kurtosis"] = numerical_data.kurt()

# Display descriptive statistics
descriptive_stats.round(3)
```

```python

# Calculate the mode of numerical variables
numerical_data.mode()
```

```python

# Frequency of asphalt mixture types
frequency = selected_data["Mix_Type"].value_counts()

# Percentage of asphalt mixture types
percentage = selected_data["Mix_Type"].value_counts(normalize=True) * 100

# Create frequency table
frequency_table = pd.DataFrame({"Frequency": frequency,"Percentage (%)": percentage})

frequency_table
```

```python
# Check negative air void values
negative_air_voids = selected_data[
    selected_data["% Air Voids"] < 0]
print("Number of negative values:", len(negative_air_voids))
negative_air_voids
```


### 8.1 Descriptive Statistics Results

The descriptive statistics were calculated for six numerical variables, each containing 180 observations.

The mean air void content was 3.076%, while the mean asphalt content was 5.500%.

The average unit weight was 2392.337 kg/m³, and the average adjusted stability was 8.058 kN.

The mean stiffness modulus was 3351.450 MPa at 20°C and 913.244 MPa at 30°C.

The four asphalt mixture types were equally represented, with 45 observations (25%) in each group.

The mode results showed that some numerical variables had multiple modes, while air voids and unit weight had no unique mode.



### 8.2 Engineering Interpretation

The mean stiffness modulus at 30°C was approximately 72.75% lower than at 20°C.

This result is consistent with the temperature-sensitive behaviour of asphalt mixtures, which generally become less stiff at higher temperatures.

Air voids, adjusted stability, and stiffness modulus showed positive skewness, while unit weight showed negative skewness.

Asphalt content had a skewness value of zero, indicating a symmetric distribution.

Further statistical tests are required to determine whether the observed differences and relationships are statistically significant.



### 8.3 Data Quality Observation

During descriptive statistical analysis, one negative air void value of -0.041606% was identified at index 27 in the HMA2 mixture group.

This observation represents approximately 0.56% of the total dataset.

Negative air void content is physically unrealistic and may be associated with measurement uncertainty or calculation error.

The original value is retained at this stage to preserve data integrity because the exact cause cannot be confirmed.

A sensitivity analysis will be considered for statistical tests involving air void content.



## 9. Data Visualisation

Data visualisation is used to explore the distributions, variability, potential outliers, and relationships between the selected asphalt mixture variables.

Histograms, boxplots, Q-Q plots, scatter plots, and a correlation heatmap are used to support exploratory data analysis and the selection of appropriate statistical tests.



### 9.1 Histograms

Histograms are used to examine the distribution of the six numerical variables.

They provide an initial visual assessment of symmetry, skewness, and potential unusual values.


```python

# Plot histograms for numerical variables
fig, axes = plt.subplots(2, 3, figsize=(16, 10))

for col, ax in zip(numerical_data.columns, axes.flatten()):
    sns.histplot(data=numerical_data,x=col,bins=15,kde=True,ax=ax,color="steelblue")
    ax.set_title(col)
    ax.set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```


#### 9.1.1 Histogram Results

The histograms show different distribution patterns across the six numerical variables.

Air voids, adjusted stability, and stiffness modulus at both temperatures exhibit positive skewness.

Unit weight exhibits slight negative skewness.

Asphalt content has five equally represented levels ranging from 4.5% to 6.5%, indicating a discrete and symmetric distribution rather than a normal distribution.

The stiffness modulus distributions show considerable variability at both temperatures.

Further assessment using Q-Q plots and normality tests is required before selecting appropriate hypothesis tests.



### 9.2 Boxplots

Boxplots are used to examine the median, interquartile range, variability, and potential outliers of the six numerical variables.

This visualisation helps identify unusual observations and understand the dispersion of asphalt mixture properties.


```python

# Create boxplots for numerical variables
fig, axes = plt.subplots(2, 3, figsize=(16, 10))

for col, ax in zip(numerical_data.columns, axes.flatten()):
    sns.boxplot(
        y=numerical_data[col],ax=ax,color="steelblue")
    ax.set_title(col)
    ax.set_ylabel(col)

plt.tight_layout()
plt.show()
```

```python

# Calculate IQR for each numerical variable
Q1 = numerical_data.quantile(0.25)
Q3 = numerical_data.quantile(0.75)
IQR = Q3 - Q1

# Calculate lower and upper boundaries
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify potential outliers
outliers = ((numerical_data < lower_bound) |(numerical_data > upper_bound))

# Create an outlier summary table
outlier_summary = pd.DataFrame({
    "Q1": Q1,
    "Q3": Q3,
    "IQR": IQR,
    "Lower Bound": lower_bound,
    "Upper Bound": upper_bound,
    "Outlier Count": outliers.sum(),
    "Outlier Percentage (%)": outliers.mean() * 100})

outlier_summary.round(3)
```

```python
# Display observations containing potential outliers
selected_data[outliers.any(axis=1)]
```


#### 9.2.2 Boxplot Results

The IQR method identified four observations containing potential outliers, representing 2.22% of the dataset.

Two lower outliers were identified in unit weight, representing 1.11% of its observations.

Two upper outliers were identified in stiffness modulus at 30°C, also representing 1.11% of its observations. Both belonged to the HMA3 mixture group.

No statistical outliers were identified in the remaining four numerical variables.

The previously identified negative air void value was not detected as an outlier by the IQR method. This highlights the importance of combining statistical analysis with engineering knowledge.

All observations were retained because the potential outliers may represent genuine material variability rather than measurement errors.



### 9.3 Q-Q Plots

Quantile-Quantile (Q-Q) plots are used to visually assess whether the numerical variables follow a normal distribution.

The observed quantiles are compared with the theoretical quantiles of a normal distribution.

Systematic deviations from the reference line indicate departures from normality.


```python
from scipy import stats

# Create Q-Q plots for numerical variables
fig, axes = plt.subplots(2, 3, figsize=(16, 10))

for col, ax in zip(numerical_data.columns, axes.flatten()):

    stats.probplot(numerical_data[col],dist="norm",plot=ax)
    ax.set_title(col)

plt.tight_layout()
plt.show()
```


#### 9.3.1 Q-Q Plot Results

The Q-Q plots indicate departures from normality across the selected numerical variables.

Air voids, adjusted stability, and stiffness modulus at both temperatures show systematic deviations from the normal reference line.

Asphalt content displays a step-like pattern because it contains five discrete levels.

Unit weight appears closer to a normal distribution than the other numerical variables, although deviations are observed in the tails.

These visual findings are consistent with the histogram and skewness results.

Formal normality testing is required to support the selection of appropriate statistical tests.



### 9.4 Scatter Plots

Scatter plots are used to investigate relationships between the physical and mechanical properties of asphalt mixtures.

The analysis focuses on the relationships between air voids, asphalt content, unit weight, adjusted stability, and stiffness modulus.

Different colours represent the four asphalt mixture types, allowing visual comparison between groups.


```python

# Select variable pairs for scatter plots
pairs = [
    ('% Air Voids', 'Unit Wt. (kg/m3)'),
    ('% A.C. by wt. (Pb)', '% Air Voids'),
    ('% Air Voids', 'ITSM (Mpa) - 20 deg'),
    ('% A.C. by wt. (Pb)', 'Adj.  Stability kN'),
    ('Adj.  Stability kN', 'ITSM (Mpa) - 20 deg'),
    ('ITSM (Mpa) - 20 deg', 'ITSM (Mpa) - 30 deg')]

# Create scatter plots
fig, axes = plt.subplots(2, 3, figsize=(18, 11))

for (x, y), ax in zip(pairs, axes.flatten()):

    sns.scatterplot(
        data=selected_data,
        x=x,
        y=y,
        hue="Mix_Type",ax=ax,alpha=0.7)

    ax.set_title(f"{x}\nvs {y}")

plt.tight_layout()
plt.show()
```


#### 9.4.1 Scatter Plot Results

The scatter plots reveal several relationships between the physical and mechanical properties of asphalt mixtures.

A strong negative relationship is observed between air void content and unit weight.

Asphalt content also shows a negative relationship with air void content.

Adjusted stability and stiffness modulus at 20°C demonstrate a strong positive relationship.

Stiffness modulus at 20°C and 30°C also shows a strong positive relationship.

No clear linear relationship is visually observed between asphalt content and adjusted stability or between air voids and stiffness modulus at 20°C.

Differences between mixture groups are also visible.

These findings are preliminary and will be further examined using correlation analysis.



### 9.5 Correlation Heatmap

Pearson correlation analysis is used to measure the strength and direction of linear relationships between the six numerical variables.

A correlation heatmap provides a visual representation of the correlation coefficients.

The results are used to identify important relationships between asphalt mixture properties and support further statistical analysis.


```python

# Calculate Pearson correlation coefficients
correlation_matrix = numerical_data.corr(method="pearson")

# Display correlation matrix
correlation_matrix.round(3)
```

```python
# Create correlation heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".3f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    center=0,
    linewidths=0.5)
plt.title("Correlation Heatmap of Asphalt Mixture Properties")
plt.tight_layout()
plt.show()
```


#### 9.5.1 Correlation Analysis Results

The strongest positive correlation was observed between stiffness modulus at 20°C and 30°C (r = 0.917).

Adjusted stability showed a very strong positive correlation with stiffness modulus at 20°C (r = 0.851) and a strong positive correlation at 30°C (r = 0.779).

Asphalt content and air void content demonstrated a very strong negative correlation (r = -0.848).

Air void content and unit weight also showed a strong negative correlation (r = -0.766).

Asphalt content and unit weight demonstrated a moderate positive correlation (r = 0.489).

In contrast, asphalt content and adjusted stability showed a very weak negative correlation (r = -0.051).

These results are consistent with the patterns observed in the scatter plots. However, correlation does not establish causation, and differences between mixture types may influence the overall relationships.

The exploratory visual analysis provided important insights into the distribution, variability and relationships of asphalt mixture properties.

The histograms and Q-Q plots indicated departures from normality in several numerical variables.

The boxplots and IQR method identified four observations containing potential outliers. Two were associated with unit weight and two with stiffness modulus at 30°C.

The scatter plots and correlation heatmap revealed important positive and negative relationships between the physical and mechanical properties of asphalt mixtures.

These findings provide a foundation for confidence interval estimation and hypothesis testing. Statistical assumptions will be assessed before selecting the appropriate tests.
