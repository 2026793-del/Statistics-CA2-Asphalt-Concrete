
# Statistical Analysis of Asphalt Concrete Mixtures
## 1. Introduction

This project investigates the mechanical properties of asphalt concrete mixtures using statistical methods. The analysis focuses on stiffness modulus, asphalt mixture type, temperature, and other material properties.

The study will use descriptive statistics, confidence intervals, and inferential statistical tests to investigate relationships and differences in the dataset.


```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load all worksheets from the Excel file
file_path = "Full Sample Data.xlsx"

sheets = pd.read_excel(file_path, sheet_name=None)

# Display worksheet names and dimensions
for name, df in sheets.items():
    print(f"{name}: {df.shape}")
```


## 2. Data Inspection

The dataset contains four worksheets: HMA2, HMA3, SMA1, and SMA2.

This section examines the column names, data types, and structure of each worksheet before data cleaning and statistical analysis.


```python

# Inspect column names and data types

for name, df in sheets.items():
    print(f"\n===== {name} =====")
    print("Shape:", df.shape)
    print("\nColumn names:")
    print(df.columns.tolist())
    print("\nData types:")
    print(df.dtypes)
```


## 3. Data Quality Assessment

This section checks all four worksheets for missing values and duplicate records.

Missing values may affect statistical calculations, while duplicate records may bias the results.

Each worksheet is checked separately before combining the datasets.


```python

# Check missing values and duplicates

for name, df in sheets.items():

    print(f"\n===== {name} =====")

    print("Total missing values:")
    print(df.isnull().sum().sum())

    print("\nMissing values by column:")
    print(df.isnull().sum())

    print("\nDuplicate rows:")
    print(df.duplicated().sum())
```


### 3.1 Data Quality Results

All four worksheets (HMA2, HMA3, SMA1, and SMA2) were checked for missing values and duplicate records.

The results showed that:

- No missing values were identified in any worksheet.
- No exact duplicate rows were found.
- Each worksheet contains 45 observations.

Therefore, no missing-value imputation or duplicate removal is required at this stage.

However, further checks of variable consistency and measurement validity are necessary before statistical analysis.



## 4. Column Consistency Assessment

Before combining the four worksheets, their column names are compared to identify any structural differences.

This step ensures that variables are aligned correctly and prevents errors during data integration.


```python

# Compare column names across worksheets

all_columns = set()

for df in sheets.values():
    all_columns.update(df.columns)

for name, df in sheets.items():
    print(f"\n===== {name} =====")

    missing_columns = all_columns - set(df.columns)

    print("Number of columns:", len(df.columns))
    print("Columns not present:", missing_columns)
```


### 4.1 Column Consistency Results

The column consistency assessment identified a structural difference between the four worksheets.

HMA2 and HMA3 each contain 16 columns, while SMA1 and SMA2 contain 15 columns.

The variable "Stab/Flow" is present in both HMA worksheets but absent from both SMA worksheets.

The remaining column names are consistent across all four worksheets.

Therefore, the 15 common variables will be used for the combined statistical analysis. The original worksheets will remain unchanged.



## 5. Dataset Integration

The four worksheets (HMA2, HMA3, SMA1, and SMA2) are combined into one dataset for statistical analysis.

A new categorical variable, `Mix_Type`, is created to identify the original asphalt mixture group.

Only the 15 common columns are retained to ensure consistency across the four worksheets. The original datasets remain unchanged.


```python

# Identify common columns
common_columns = list(sheets[next(iter(sheets))].columns)

for df in sheets.values():
    common_columns = [col for col in common_columns if col in df.columns]

# Combine worksheets and add mixture type
combined_data = pd.concat([df[common_columns].assign(Mix_Type=name) for name, df in sheets.items()],ignore_index=True)

# Check the combined dataset
print("Combined dataset shape:", combined_data.shape)

print("\nObservations by mixture type:")
print(combined_data["Mix_Type"].value_counts())

print("\nFirst five rows:")
display(combined_data.head())
```

### 5.1 Dataset Integration Results

The four worksheets were successfully combined into a single dataset containing 180 observations and 16 variables.

The combined dataset contains 15 numerical variables and one categorical variable, Mix_Type.

Each mixture type (HMA2, HMA3, SMA1, and SMA2) contains 45 observations, representing 25% of the total dataset.

The Stab/Flow variable was excluded because it was not available in the SMA worksheets.

The original worksheets remain unchanged.

## 8. Descriptive Statistics

Descriptive statistics are used to summarise the numerical characteristics of the selected asphalt mixture variables.

Measures of central tendency, dispersion, and distribution are calculated to understand the physical and mechanical properties of asphalt mixtures.

These statistics provide an initial understanding of the dataset before confidence interval estimation and hypothesis testing.


```python

# Select numerical variables
numerical_data = selected_data.select_dtypes(include=np.number)

# Calculate descriptive statistics
descriptive_stats = numerical_data.describe().T

# Calculate additional statistical measures
descriptive_stats["median"] = numerical_data.median()
descriptive_stats["variance"] = numerical_data.var()
descriptive_stats["range"] = numerical_data.max() - numerical_data.min()
descriptive_stats["IQR"] = numerical_data.quantile(0.75) - numerical_data.quantile(0.25)
descriptive_stats["skewness"] = numerical_data.skew()
descriptive_stats["kurtosis"] = numerical_data.kurt()

# Display descriptive statistics
descriptive_stats.round(3)
```

```python

# Calculate the mode of numerical variables
numerical_data.mode()
```

```python

# Frequency of asphalt mixture types
frequency = selected_data["Mix_Type"].value_counts()

# Percentage of asphalt mixture types
percentage = selected_data["Mix_Type"].value_counts(normalize=True) * 100

# Create frequency table
frequency_table = pd.DataFrame({"Frequency": frequency,"Percentage (%)": percentage})

frequency_table
```


### 8.1 Descriptive Statistics Results

The descriptive statistics were calculated for six numerical variables, each containing 180 observations.

The mean air void content was 3.076%, while the mean asphalt content was 5.500%.

The average unit weight was 2392.337 kg/m³, and the average adjusted stability was 8.058 kN.

The mean stiffness modulus was 3351.450 MPa at 20°C and 913.244 MPa at 30°C.

The four asphalt mixture types were equally represented, with 45 observations (25%) in each group.

The mode results showed that some numerical variables had multiple modes, while air voids and unit weight had no unique mode.



### 8.2 Engineering Interpretation

The mean stiffness modulus at 30°C was approximately 72.75% lower than at 20°C.

This result is consistent with the temperature-sensitive behaviour of asphalt mixtures, which generally become less stiff at higher temperatures.

Air voids, adjusted stability, and stiffness modulus showed positive skewness, while unit weight showed negative skewness.

Asphalt content had a skewness value of zero, indicating a symmetric distribution.

Further statistical tests are required to determine whether the observed differences and relationships are statistically significant.



### 8.3 Data Quality Observation

During descriptive statistical analysis, one negative air void value of -0.041606% was identified at index 27 in the HMA2 mixture group.

This observation represents approximately 0.56% of the total dataset.

Negative air void content is physically unrealistic and may be associated with measurement uncertainty or calculation error.

The original value is retained at this stage to preserve data integrity because the exact cause cannot be confirmed.

A sensitivity analysis will be considered for statistical tests involving air void content.
